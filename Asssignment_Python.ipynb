```python
#Question 8.3
counter = 0

def increment_counter():
    counter = counter + 1

increment_counter()
#inside the function, it treats counter as a local variable because you're assigning to it.But the local counter hasn't been given a value yet.
```

```python
#Question 8.4
counter = 0

def increment_counter():
    global counter
    counter = counter + 1

increment_counter()
print(counter)
```
