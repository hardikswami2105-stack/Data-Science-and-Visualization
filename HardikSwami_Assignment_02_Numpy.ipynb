## Question 35

```python
rng = np.random.default_rng(2026)
marks = rng.integers(20, 101, size=(12, 4))
averages = marks.mean(axis=1)
std_devs = marks.std(axis=1)
pass_counts = (marks >= 40).sum(axis=1)
passing_all = np.where(pass_counts == 4)[0] + 1
average_75 = np.where(averages >= 75)[0] + 1

print("Marks (rows = students 1–12):\n", marks)
print("Averages:", np.round(averages, 2))
print("Standard deviations:", np.round(std_devs, 2))
print("Tests passed per student:", pass_counts)
print("Students passing all four tests:", passing_all)
print("Students with average >= 75:", average_75)
print("Student with highest average:", int(np.argmax(averages) + 1),
      "with average", round(float(averages.max()), 2))
```

## Question 37

```python
quantities = np.array([
    [150, 240, 90, 320],
    [180, 200, 120, 300],
    [210, 150, 100, 280],
    [170, 260, 110, 310],
    [190, 220, 130, 290],
    [200, 250, 140, 330]
])
prices = np.array([150, 240, 90, 320])
revenue = quantities * prices  # broadcast prices across six days
daily_revenue = revenue.sum(axis=1)
product_revenue = revenue.sum(axis=0)
discounted_daily_revenue = np.where(daily_revenue > 1000, daily_revenue * 0.90, daily_revenue)

print("Revenue matrix:\n", revenue)
print("Daily revenue:", daily_revenue)
print("Total revenue per product:", product_revenue)
print("Highest-revenue day:", int(daily_revenue.argmax() + 1))
print("Highest daily revenue:", daily_revenue.max())
print("Best-selling product by revenue (column 1–4):", int(product_revenue.argmax() + 1))
print("Daily revenue after applicable 10% discount:", np.round(discounted_daily_revenue, 2))
print("Total revenue after discounts:", round(float(discounted_daily_revenue.sum()), 2))
```

## Question 38

```python
rng = np.random.default_rng(38)
original = rng.normal(loc=50, scale=12, size=100)
cleaned = np.clip(original, 30, 70)

def describe(x):
    return {
        "mean": round(float(np.mean(x)), 3),
        "variance": round(float(np.var(x)), 3),
        "standard deviation": round(float(np.std(x)), 3),
        "minimum": round(float(np.min(x)), 3),
        "maximum": round(float(np.max(x)), 3),
    }

print("Original statistics:", describe(original))
print("Cleaned statistics:", describe(cleaned))
print("Number of values changed:", int(np.count_nonzero(original != cleaned)))
print("Boundary cleaning limits extreme values to [30, 70]. This generally reduces variance and standard deviation; the mean may also shift.")
```

## Question 39

```python
branch_a = pd.Series([120, 0, 145, 180], index=["Q1", "Q2", "Q3", "Q4"])
branch_b = pd.Series([0, 110, 0, 155], index=["Q1", "Q2", "Q3", "Q4"])
branch_c = pd.Series([95, 0, 125, 140], index=["Q1", "Q3", "Q3", "Q4"])

print("Branch A:\n", branch_a)
print("Branch B:\n", branch_b)
print("Branch C (contains duplicate Q3 label):\n", branch_c)
print("\nOrdinary addition A + B:")
print(branch_a + branch_b)
print("\nCommon quarter labels across all branches:")
common = set(branch_a.index) & set(branch_b.index) & set(branch_c.index)
print(sorted(common))
print("\nNote: branch_c has duplicate Q3, so arithmetic with it may expand/alignment-match duplicate labels.")
print("Separate zero-filled combined earnings (A + B; all quarter labels aligned):")
combined_zero = branch_a.add(branch_b, fill_value=0)
print(combined_zero)
```

## Question 40

```python
scores = pd.Series([78, 65, 92, 95, 48],
                    index=["R101", "R102", "R104", "R105", "R108"])
required = ["R105", "R103", "R101", "R106", "R104", "R102"]
s = scores.reindex(required)

print("Required order, with missing records as NaN:\n", s)
print("Missing student records:", s[s.isna()].index.tolist())
valid_scores = s[s.notna() & (s >= 70)]
print("Valid scores >= 70:\n", valid_scores)
result_dict = valid_scores.to_dict()
print("Dictionary:", result_dict)
print("Dictionary labels match filtered Series:", list(result_dict.keys()) == list(valid_scores.index))
```
