### Q26. Create course_df and perform the requested sequence

```python
course_df = pd.DataFrame({
    'Student': ['Aman', 'Riya', 'Karan', 'Neha', 'Vikas'],
    'Course': ['Python', 'Java', 'C', 'Python', 'Java'],
    'Score': [88, 92, 79, 85, 90]
})

print(course_df.columns)
print(course_df['Student'])

course_student_score_df = course_df[['Student', 'Score']]
print(course_student_score_df)

course_df['Batch'] = '2026'
course_df
```

### Q27. Add np.arange(5), then replace with a partially indexed Series

```python
travel_df = pd.DataFrame({
    'Student': ['Aman', 'Riya', 'Karan', 'Neha', 'Vikas'],
    'Destination': ['Delhi', 'Agra', 'Jaipur', 'Goa', 'Manali']
})

travel_df['Code'] = np.arange(5)
print("After np.arange(5):")
print(travel_df)

travel_codes = pd.Series([101, 104], index=[1, 4])
travel_df['Code'] = travel_codes
print("\nAfter assigning the partially indexed Series:")
print(travel_df)
```

### Q28. Create school_df, request missing House, then fill House

```python
school_df = pd.DataFrame({
    'Student': ['Aman', 'Riya', 'Karan', 'Neha'],
    'Class': [10, 10, 11, 11],
    'Section': ['A', 'B', 'A', 'B']
})

school_with_house = school_df.reindex(
    columns=['Student', 'Class', 'Section', 'House']
)
print("Before assigning House:")
print(school_with_house)

school_with_house['House'] = 'Tagore'
print("\nAfter assigning a common House value:")
print(school_with_house)
```

### Q29. Add a partially indexed Series, then delete it

```python
inventory_df = pd.DataFrame({
    'Item': ['Pen', 'Book', 'Bag', 'Pencil', 'Notebook'],
    'Quantity': [20, 15, 8, 30, 12]
})

stock_status = pd.Series(
    ['Available', 'Low', 'Available'],
    index=[0, 2, 4]
)
inventory_df['Status'] = stock_status

print("After adding Status:")
print(inventory_df)

inventory_df = inventory_df.drop(columns='Status')
print("\nAfter deleting Status:")
print(inventory_df)
```

### Q30. Complete sequence using a copied table

```python
# Copy a table with at least four columns before running this cell.
lab_df = pd.read_clipboard()

print("Complete DataFrame:")
print(lab_df)

print("\nColumn names:")
print(lab_df.columns)

# Retrieve a single-word column (first column)
single_word_column = lab_df[lab_df.columns[0]]
print("\nSingle-word column:")
print(single_word_column)

# Retrieve a column containing a space, if present
space_columns = [c for c in lab_df.columns if ' ' in str(c)]
if space_columns:
    print("\nColumn containing a space:")
    print(lab_df[space_columns[0]])
else:
    print("\nNo column containing a space was found in the copied table.")

# Construct a smaller DataFrame from selected columns
smaller_df = lab_df.iloc[:, :3]
print("\nSmaller DataFrame:")
print(smaller_df)

# Add a new whole column
lab_df['NewColumn'] = 'Added'
print("\nAfter adding NewColumn:")
print(lab_df)

# Delete the added column
lab_df = lab_df.drop(columns='NewColumn')
print("\nAfter deleting NewColumn:")
print(lab_df)
```
