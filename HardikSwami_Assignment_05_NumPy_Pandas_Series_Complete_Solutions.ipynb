## Question 35

```python
rng = np.random.default_rng(42)
marks = rng.integers(20, 101, size=(12, 4))
averages = marks.mean(axis=1)
std_devs = marks.std(axis=1)
pass_all = np.all(marks >= 35, axis=1)
test_std_devs = marks.std(axis=0)
print('Marks (12 students × 4 tests):\n', marks)
print('Student averages:', np.round(averages, 2))
print('Student standard deviations:', np.round(std_devs, 2))
print('Students passing all tests:', int(pass_all.sum()))
print('Student with highest average (1-based):', int(np.argmax(averages) + 1))
print('Test standard deviations:', np.round(test_std_devs, 2))
print('Test with greatest variation (1-based):', int(np.argmax(test_std_devs) + 1))
```

## Question 36

```python
matrix = np.arange(21, 46).reshape(5, 5)
central = matrix[1:4, 1:4]
transposed = central.T
product = central @ transposed
row_sums = product.sum(axis=1)
print('Original matrix:\n', matrix)
print('Central 3×3 matrix:\n', central)
print('Transpose:\n', transposed)
print('Product:\n', product)
print('Product row sums:', row_sums)
print('Maximum row sum:', row_sums.max())
print('Position (1-based) of maximum row sum:', int(np.argmax(row_sums) + 1))
```

## Question 37

```python
quantities = np.array([
    [20, 15, 30, 25],
    [18, 22, 28, 20],
    [25, 20, 35, 18],
    [22, 18, 32, 24],
    [30, 25, 20, 28],
    [24, 30, 26, 22]
])
unit_prices = np.array([150, 240, 90, 320])
product_quantities = quantities.sum(axis=0)
revenue_by_product = product_quantities * unit_prices
print('Total quantity per product:', product_quantities)
print('Revenue per product:', revenue_by_product)
print('Total revenue:', revenue_by_product.sum())
print('Highest-revenue product (1-based):', int(np.argmax(revenue_by_product) + 1))
print('Best-selling product by quantity (1-based):', int(np.argmax(product_quantities) + 1))
discounted_prices = np.where(product_quantities > 100, unit_prices * 0.90, unit_prices)
discounted_revenue = product_quantities * discounted_prices
print('Adjusted unit prices:', discounted_prices)
print('Revenue after discounts per product:', discounted_revenue)
print('Total revenue after discounts:', discounted_revenue.sum())
```

## Question 38

```python
rng = np.random.default_rng(42)
data = rng.normal(loc=50, scale=12, size=100)
outside = (data < 30) | (data > 70)
cleaned = data.copy()
cleaned[outside] = np.median(data)
def stats(x):
    return {'mean': x.mean(), 'variance': x.var(), 'std': x.std(),
            'min': x.min(), 'max': x.max()}
print('Observations outside [30, 70]:', int(outside.sum()))
print('Original statistics:', stats(data))
print('Cleaned statistics:', stats(cleaned))
print('Note: boundary replacement changes the distribution; the median used is from the original data.')
```

## Question 39

```python
branch_a = pd.Series({'Q1': 120, 'Q2': 145, 'Q4': 180})
branch_b = pd.Series({'Q2': 110, 'Q3': 135, 'Q4': 155})
branch_c = pd.Series({'Q1': 95, 'Q3': 125, 'Q4': 140})
ordinary_sum = branch_a + branch_b + branch_c
print('Ordinary addition (NaN where a quarter is missing in any branch):\n', ordinary_sum)
print('Non-missing ordinary totals:\n', ordinary_sum.dropna())
combined_zero = branch_a.add(branch_b, fill_value=0).add(branch_c, fill_value=0)
print('Combined earnings, treating absent records as 0:\n', combined_zero)
all_three = pd.concat([branch_a.rename('A'), branch_b.rename('B'), branch_c.rename('C')], axis=1)
all_three['All three recorded'] = all_three.notna().all(axis=1)
print('Quarter-by-branch table:\n', all_three)
print('Quarters recorded in all branches:', all_three.index[all_three['All three recorded']].tolist())
```
